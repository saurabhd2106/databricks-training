# Actuarial Claims Lakehouse Pipeline

This pipeline is split into five stage notebooks under `src/notebooks/`.
The DAB job `actuarial_claims_job` runs them in order:

| Task | Notebook |
|---|---|
| `setup` | [`notebooks/01_setup.ipynb`](notebooks/01_setup.ipynb) |
| `bronze` | [`notebooks/02_bronze_ingest.ipynb`](notebooks/02_bronze_ingest.ipynb) |
| `data_quality` | [`notebooks/03_data_quality.ipynb`](notebooks/03_data_quality.ipynb) |
| `silver` | [`notebooks/04_silver.ipynb`](notebooks/04_silver.ipynb) |
| `gold` | [`notebooks/05_gold.ipynb`](notebooks/05_gold.ipynb) |

Run via:
```bash
databricks bundle run actuarial_claims_job --target dev
```
